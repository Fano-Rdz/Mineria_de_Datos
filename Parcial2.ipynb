```python
dfc = pd.read_csv('clientes.csv')
dfo = pd.read_csv('ordenes.csv')
dfc
```

```python
df_merged = pd.merge(dfc, dfo, on='cliente_id', how='left')
df_merged
```

```python
total_rows_merged = len(df_merged)
total_monto_por_cliente = df_merged.groupby('cliente_id')['monto'].sum().fillna(0).reset_index()
total_monto_por_cliente.rename(columns={'monto': 'monto_total'}, inplace=True)
display(total_monto_por_cliente)
```

```python
df = pd.read_csv('sensores.csv')
df
```

```python
import pandas as pd
df = df[['ts', 'temperatura', 'humedad']]

df['ts'] = pd.to_datetime(df['ts'], format='mixed', errors='coerce')

mediana_temperatura = df['temperatura'].median()
df['temperatura'] = df['temperatura'].fillna(mediana_temperatura)

media_humedad = df['humedad'].mean()
df['humedad'] = df['humedad'].fillna(media_humedad)

nat_ts = df['ts'].isnull().sum()
print(f"(a) Número de NaT restantes en 'ts': {nat_ts}")

final_mediana_temperatura = round(df['temperatura'].median(), 1)
print(f"(b) Mediana final de 'temperatura' tras imputación: {final_mediana_temperatura}")

final_media_humedad = round(df['humedad'].mean(), 3)
print(f"(c) Media final de 'humedad' tras imputación: {final_media_humedad}")
```
